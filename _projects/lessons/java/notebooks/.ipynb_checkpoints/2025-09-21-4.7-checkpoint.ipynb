## Learning Objectives
- Define **class variables** using `static`
- Define **class methods** using `static`
- Understand that class variables are shared by all objects
- Understand what class methods can and cannot access
- Use **class name** and **dot operator** to access public class variables/methods
- Understand the `final` keyword

## Instance vs. Class Variables

### Instance Variables

Recall that instance variables belong to an individual object:
```java
public Class Student {
    private String name;
    private int grade;
}
```

So, if we create 3 students:
```java
Student s1 = new Student();
Student s2 = new Student();
Student s3 = new Student();
```
Each object gets *its own `name` and `grade`*.

### Class Variables

...belong to *the clas itself* rather than individual objects, which we define using `static`:
```java
public class Student {
    private String name;
    private int grade;

    public static int numberOfStudents = 0;
}
```

There is only *one copy* of `numberOfStudents`, shared by every `Student` object.

**`static` means that the variable or methods belongs to the class rather than each individual object**.

## Public Class Variables

If we define a class variable as `public`, it can be accessed outside of the class using
```java
ClassName.variableName
```
like this:
```java
public class Game {
    public static int players = 0;
}
```

Then, we could do
```java
System.out.println(Game.players); // get players
Game.players = 5; // modify players
```

## Class Methods

Similarly, a class method is declared via `static`:
```java
public class MathHelper {

    public static int doubleValue(int x) {
        return x * 2;
    }
}
```

In which case we can call it *without* creating an object:
```java
int result = MathHelper.doubleValue(5);

System.out.println(result);
```

You can also call class methods within a class method:
```java
public class Calculator {

    public static int square(int x) {
        return x * x;
    }

    public static int squareAndAddOne(int x) {
        return square(x) + 1;
    }
}
```

## `final` Keyword

We define a variable as `final` if we want it to be immutable:
```java
public static final int DAYS_IN_WEEK = 7;
```
